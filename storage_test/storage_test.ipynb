```python
import pandas as pd
from battery_storage_eff import BatteryStorageEff

df = pd.DataFrame({
    'balance': [100, -100, 50, -50]  # in kW
})

# Mit Effizienz von 95% beim Laden und Entladen
storage = BatteryStorageEff(
    max_capacity_kwh=500, #235000
    initial_charge_kwh=250,
    charge_efficiency=0.95,
    discharge_efficiency=0.95
)

result = storage.process_balance(df['balance'], timestep_hours=1.0)

print(result)
print("\nStatistiken:")
print(storage.get_stats(result))
```
